````python
#4. Write a function `safe_downcast(arr, target_dtype)` that:
   #- Checks if the values fit in the target dtype range
   #- If yes: casts and returns the result
   #- If no: prints a warning and returns the original array


def safe_downcast(arr, target_dtype):
    
    dtype_info = np.iinfo(target_dtype)

    min_allowed = dtype_info.mint7
    max_allowed = dtype_info.max

    if arr.min() >= min_allowed and arr.max() <= max_allowed:
        return arr.astype(target_dtype)

    print("Warning: Values do not fit in", target_dtype)
    return arr
````

## Assignment 6 — NaN & Inf Handling (Data Cleaning)

Given sensor data from an IoT device:
```python
sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])
```

1. Count the number of `NaN` values.
2. Count the number of `Inf` values (both +inf and -inf).
3. Extract only the valid (finite) readings.
4. Replace NaN values with the **mean** of all finite values.
5. Replace Inf values with the **max** finite value and -Inf with the **min** finite value.
6. Calculate `nanmean`, `nanmin`, `nanmax`, and `nanstd` of the sensor array.
7. **Bonus**: Write a single function `clean_sensor_data(arr)` that handles all of the above and returns a clean array.

````python
sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])

#1. Count the number of `NaN` values.
NaN_count = np.sum(np.isnan(sensor))
print(NaN_count)

#2. Count the number of `Inf` values (both +inf and -inf).
no_infinite = np.sum(np.isinf(sensor))
print(no_infinite)

#3. Extract only the valid (finite) readings.

clean_data = sensor[np.isfinite(sensor)]
print(clean_data)

#4. Replace NaN values with the **mean** of all finite values.
filled_mean = sensor.copy()

mean_val = np.mean(sensor[np.isfinite(sensor)])

filled_mean[np.isnan(filled_mean)] = mean_val

print(filled_mean)

````

````python
#5. Replace Inf values with the **max** finite value and -Inf with the **min** 
#finite value.
sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])

filled = sensor.copy()

finite_vals = filled[np.isfinite(filled)]

max_val = np.max(finite_vals)
min_val = np.min(finite_vals)

filled[np.isposinf(filled)] = max_val
filled[np.isneginf(filled)] = min_val

print(filled)
````

````python
#6. Calculate `nanmean`, `nanmin`, `nanmax`, and `nanstd` of the sensor array.
print("Mean:", np.nanmean(sensor))
print("Min :", np.nanmin(sensor))
print("Max :", np.nanmax(sensor))
print("Std :", np.nanstd(sensor))

````

````python
#7. **Bonus**: Write a single function `clean_sensor_data(arr)` that handles 
# all of the above and returns a clean array.

sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])

def clean_sensor_data(arr):
    cleaned = arr.copy()

    # Find finite values
    finite_vals = cleaned[np.isfinite(cleaned)]

    # Min, max, and mean of finite values
    max_val = np.max(finite_vals)
    min_val = np.min(finite_vals)

    # Replace infinities
    cleaned[np.isposinf(cleaned)] = max_val
    cleaned[np.isneginf(cleaned)] = min_val

    # Recalculate mean after Inf replacement
    mean_val = np.mean(cleaned[np.isfinite(cleaned)])

    # Replace NaNs
    cleaned[np.isnan(cleaned)] = mean_val

    return cleaned

clean_sensor_data(sensor)
````
