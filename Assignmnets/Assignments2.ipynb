> **Real-world note:** In deep learning, models store millions of weights as `float32`.  
> Knowing `.nbytes` helps you estimate whether a model fits in GPU memory!

```python
# **Q1.** Create a 2D array of shape `(4, 5)` using `np.arange(20).reshape(4, 5)`.  
# Print all six attributes: `.ndim`, `.shape`, `.size`, `.dtype`, `.itemsize`, `.nbytes`.

arr = np.arange(20).reshape(4, 5)

print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
print(arr.nbytes)
```

```python
# **Q2.** Create two arrays with the **same values** but different dtypes: `float64` and `float32`.  
# Compare their `.itemsize` and `.nbytes`. How much memory do you save with `float32`?

arr = np.array([1,2,3], dtype= np.float64)
arr1 = np.array([1,2,3], dtype= np.float32)

print(arr.itemsize)
print(arr1.itemsize)

print(arr.nbytes)
print(arr1.nbytes)

```
